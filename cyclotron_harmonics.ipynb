```python
from astropy import units as u
from astropy import constants as const
import numpy as np
```

Here are the cyclotron harmonics and orders from Parsons et al. 2021 for SDSS J2229+1853.

```python
n2_wave=6400 *u.angstrom
n3_wave=4250*u.angstrom
n4_wave=3200*u.angstrom
```

```python
def wave_to_freq(wave):
    return (const.c/wave).to(u.Hz)
```

```python
def freq_to_wave(freq):
    return (const.c/freq).to(u.angstrom)
```

```python
def get_B_field(freq):
    print('freq',freq)
    return (2 * np.pi * const.m_e * freq/ (const.e).to(u.C)).to(u.MG)
```

```python
get_B_field(44555.3*u.Hz)
```

```python
def get_harmonics(wave,guessed_n):
    test_list=[1,2,3,4]
    test_list.remove(guessed_n)
    freq=wave_to_freq(wave)
    other_freqs=np.array(test_list)/float(guessed_n)*freq
    #print(test_list,"other_freqs",other_freqs)
    if guessed_n == 1:
        fundamental_freq=freq
    else:
        fundamental_freq=other_freqs[np.where(np.array(test_list)==1)]
    mag_field=get_B_field(fundamental_freq)
    print('B field:',mag_field,mag_field.to(u.T))
    other_waves=freq_to_wave(other_freqs)
    print(test_list,"other_waves",other_waves)
    return
```

```python
j1113_peak=3900 * u.angstrom
```

```python
for guessed_n in [1,2,3,4]:
    print('\n\n----')
    print('guessed_n',guessed_n)
    get_harmonics(j1113_peak, guessed_n)
    print('\n\n----')
```

```python
for guessed_n in [1,2,3,4]:
    print('\n\n----')
    print('guessed_n',guessed_n)
    get_harmonics(n2_wave,guessed_n)
    print('\n\n----')
```

From page 686 of B.O.B., cyclotron frequency is given by:

$$ \nu_c = \frac{e B}{2 \pi m_e} $$

So solving that back for the magnetic field, we get

$$ B = \frac{2 \pi m_e \nu_c}{e} $$
