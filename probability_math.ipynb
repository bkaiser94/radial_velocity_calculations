Doing V_T math for the DAHes. There are currently 28 confirmed ones according to Berry, so 

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def nfunc(p):
    return -2/np.log10(p)
```

```python
p_range=np.arange(0.01,1,0.01)
n_range=nfunc(p_range)
```

```python
plt.plot(p_range,n_range)
plt.xlabel('Probability of single draw <60 km/s')
plt.ylabel('Number of WDs needed in sample')
plt.show()
```

So to put numbers on that... For each of the following probabilities of drawing a single WD with V_T <60km/s, we'd need the subsequent numbers of WDs for our probability of the whole sample having V_T <60km/s to be >3-sigma as a result.

```python
short_p=np.arange(0.7,1.0,0.025)
for p in short_p:
    print(p,nfunc(p))
```
