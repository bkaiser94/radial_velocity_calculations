# Generate log(Ca/H) vs. Teff plot (kind of like Blouin & Xu 2022)

```python
from __future__ import print_function



import matplotlib
matplotlib.use('pdf')



import numpy as np
import matplotlib.pyplot as plt
import sys
import os
from astropy.io import fits
from glob import glob
from astropy.time import Time
from astropy import coordinates as coords
from astropy import units as u
from astropy import constants as const
from astropy import convolution as conv
from astropy.table import Table, Column
import scipy.interpolate as scinterp
import time
start = time.time()

sys.path.append('../')



import spec_plot_tools as spt
import cal_params as cp

import plot_spec as ps

print(os.getcwd())

#plt.rc('lines',linewidth=0.5)
```

```python
target_dir='/Users/BenKaiser/Desktop/Kdwarf_impostor_paper'
input_file='DZ_hdominated_abundances.csv'
os.chdir(target_dir)
```

```python
figure_output_dir='/Users/BenKaiser/Desktop/Kdwarf_impostor_paper/figures/'
```

```python
input_table=Table.read(input_file)
```

```python
solar_cah=6.32-12 #from Lodders 2019 A(Ca)-12
sol_offsets=[1,0,-1,-2]
```

```python
fig=spt.start_ApJ_fig(width_cols=1,width_height=[1.,6./10.])
blouinxu=np.where(input_table['source']=='blouinandxu2022')
blouinxu_table=input_table[blouinxu]
obrien=np.where(input_table['source']=='obrien2022')
obrien_table=input_table[obrien]
plt.scatter(blouinxu_table['teff'],blouinxu_table['ca/h'],marker='o',color='grey',s=6)
plt.scatter(obrien_table['teff'],obrien_table['ca/h'],marker='s',color='grey',s=6)
input_table.pprint()




J0212_row=input_table[np.where(input_table['name']=='WD J0212-5522')]
#plt.scatter(input_table['teff'],input_table['ca/h'],marker='o',color='grey',s=6)
plt.scatter(J0212_row['teff'],J0212_row['ca/h'],marker='*',s=20, color='r',label='WD J0212–5522')
plt.xlabel(r'$T_{\rm{eff}}$ (K)')
plt.ylabel('log(Ca/H)')
for off in sol_offsets:
    if off==0:
        plt.text(5000,solar_cah+off,'[Ca/H]=0')
        plt.axhline(y=solar_cah+off, color='k',linestyle='--')
    else:
        plt.axhline(y=solar_cah+off, color='k',linestyle=':')
plt.xlim([8500,3500])
#plt.legend()

print(os.getcwd())
os.chdir(figure_output_dir)
print(os.getcwd())
start = time.time()
print(start)
time_string=str(start).split('.')[0]
plt.savefig('cah_vs_teff_'+time_string+'.pdf')#plt.grid(True)



plt.show()
```
